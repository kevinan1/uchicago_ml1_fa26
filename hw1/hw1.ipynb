2) The cross entropy loss is given by:
$$
L(Y, f(X)) = - \left(Y \log f(X) + (1 - Y)\log(1 - f(X))\right)
$$

We then have 
$$
\operatorname*{arg\,min}_{f} \mathbb{E}_{X, Y} [L(Y, f(X))]
$$

By iterated expectations, we have 
$$
\mathbb{E}_{X, Y} [L(Y, f(X))] = \mathbb{E}_X \left[\mathbb{E}_{Y | X} [L(Y, f(X))]\right]
$$

We have
$$
\mathbb{E}_{Y | X = x} [L(Y, f(X))] = -\Pr(Y=1 | X=x)\log(f(x)) - \Pr(Y=0|X=x)\log(1-f(x))
$$

To minimize the expectation of the conditional, since $f$ is unrestricted, we can take $f(x) = t$ to be the value $t$ which minimizes the conditional pointwise. We have 

$$
g(t) = -\Pr(Y=1 | X=x)\log(t) - \Pr(Y=0|X=x)\log(1-t)
$$

then 
$$
g'(t) = \frac{-\Pr(Y=1 | X=x)}{t} + \frac{\Pr(Y=0 | X=x)}{1-t} = \frac{-\Pr(Y=1 | X=x)}{t} + \frac{1 - \Pr(Y=1 | X=x)}{1-t}
$$

Setting to 0 and solving, we get

$$
t = \Pr(Y=1 | X=x) 
$$

Since $g'(t)$ is negative when $t < \Pr(Y=1 | X=x)$ and positive when $t > \Pr(Y=1 | X=x)$, g attains a minimum at this point. 
Minimizing the conditional expectation pointwise minimizes the expectation of the conditional, so we have $f^*(x) = \Pr(Y=1 | X=x)$ and the claim follows.
